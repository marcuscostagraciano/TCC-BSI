```python
import pandas as pd
import matplotlib.pyplot as plt

import os
from pathlib import Path
```

```python
def read_csv(file_path: Path) -> pd.DataFrame:
	"""
	Reads a CSV file and returns a DataFrame.
	
	:param file_path: Path to the CSV file.
	:return: DataFrame containing the CSV data.
	"""
	try:
		df = pd.read_csv(file_path)
		return df
	except Exception as e:
		print(f"Error reading {file_path}: {e}")
		return None
```

```python
DATA_FOLDER = './data/'

for file in os.listdir(Path(DATA_FOLDER)):
	if (fileName := DATA_FOLDER + file).endswith('.csv'):
		df = read_csv(Path(fileName))

		plt_return = plt.violinplot(
			df,
			showextrema=True,
			showmeans=True,
			positions=(positions := range(1, len(df.columns) + 1))
		)

		plt.title(file[:-4])
		plt.ylabel('Tempo (ms)')
		plt.xticks(positions, labels=df.columns)

		plt_return['bodies'][0].set_facecolor('#64a4cc')
		plt_return['bodies'][1].set_facecolor('#fb9230')
		plt.legend(df.columns, loc="upper right")

		plt.grid()
		plt.show()
```
